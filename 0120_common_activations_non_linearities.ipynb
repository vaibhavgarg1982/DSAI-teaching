```python
import torch
import matplotlib.pyplot as plt
```

```python
x= torch.linspace(-5, 5, 100, requires_grad=False)
```

```python
def plot(x,y, ylim=(-1.1, 1.1)):
    plt.plot(x.numpy(), y.numpy())
    # plot x and y axes
    plt.axhline(0, color='black', lw=2)
    plt.axvline(0, color='black', lw=2)

    plt.grid()
    plt.xticks(torch.arange(-5, 6, 1))
    plt.xlim(-5, 5)
    plt.ylim(ylim)
    plt.show()
```

```python
sigma = torch.nn.Sigmoid()
plot(x, sigma(x))
```

```python
th = torch.nn.Tanh()
plot(x, th(x))
```

```python
ReLU = torch.nn.ReLU()
plot(x, ReLU(x), ylim=(-5, 5))
```

```python
# lin = torch.nn.Linear(1, 1)
plot(x, x, ylim=(-5, 5))
```

```python
elu = torch.nn.ELU()
plot(x, elu(x), ylim=(-5, 5))
```

```python
leakyrelu = torch.nn.LeakyReLU(negative_slope=0.1)
plot(x, leakyrelu(x), ylim=(-5, 5))
```
