```python
# convert the ewma to a function, that takes a np array and a lambda value
def ewma(ser, lambda_, bias_correction=True):
    ewm_value = 0  # Initial state
    ewm_ser = []
    for i in range(len(ser)):
        ewm_value = lambda_ * ewm_value + (1-lambda_) * ser[i]
        # bias correction
        if bias_correction:
            bias_corrected_val = ewm_value / (1 - lambda_**(i+1))
        else:
            bias_corrected_val = ewm_value
        ewm_ser.append(bias_corrected_val)
    return np.array(ewm_ser)
```
